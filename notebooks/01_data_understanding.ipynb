```python
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"]
)

shipping_features = (
    order_items
    .merge(
        orders[["order_id", "order_purchase_timestamp"]],
        on="order_id",
        how="left"
    )
)

shipping_features["order_purchase_timestamp"] = pd.to_datetime(
    shipping_features["order_purchase_timestamp"]
)

shipping_features["shipping_window_hours"] = (
    shipping_features["shipping_limit_date"]
    - shipping_features["order_purchase_timestamp"]
).dt.total_seconds() / 3600

shipping_features = (
    shipping_features
    .groupby("order_id")
    .agg(
        avg_shipping_window_hours=("shipping_window_hours", "mean"),
        min_shipping_window_hours=("shipping_window_hours", "min")
    )
    .reset_index()
)

shipping_features.head()
```

```python
geo_lookup = (
    geolocation
    .groupby("geolocation_zip_code_prefix")
    .agg(
        latitude=("geolocation_lat", "mean"),
        longitude=("geolocation_lng", "mean")
    )
    .reset_index()
)

geo_lookup.head()
```

```python
print("Unique ZIP prefixes:", len(geo_lookup))
```

```python
geo_orders = (
    order_items[
        ["order_id", "seller_id"]
    ]
    .merge(
        order_base[
            [
                "order_id",
                "customer_zip_code_prefix"
            ]
        ],
        on="order_id",
        how="left"
    )
    .merge(
        sellers[
            [
                "seller_id",
                "seller_zip_code_prefix"
            ]
        ],
        on="seller_id",
        how="left"
    )
)

geo_orders.head()
```

```python
geo_orders = geo_orders.merge(
    geo_lookup,
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left"
)

geo_orders = geo_orders.rename(
    columns={
        "latitude": "customer_lat",
        "longitude": "customer_lng"
    }
)

geo_orders = geo_orders.drop(
    columns=["geolocation_zip_code_prefix"]
)

geo_orders = geo_orders.merge(
    geo_lookup,
    left_on="seller_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left"
)

geo_orders = geo_orders.rename(
    columns={
        "latitude": "seller_lat",
        "longitude": "seller_lng"
    }
)

geo_orders = geo_orders.drop(
    columns=["geolocation_zip_code_prefix"]
)

geo_orders.head()
```

```python
from numpy import radians, sin, cos, sqrt, arctan2

lat1 = radians(geo_orders["customer_lat"])
lon1 = radians(geo_orders["customer_lng"])
lat2 = radians(geo_orders["seller_lat"])
lon2 = radians(geo_orders["seller_lng"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    sin(dlat / 2) ** 2
    + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
)

geo_orders["seller_customer_distance_km"] = (
    2 * 6371 * arctan2(sqrt(a), sqrt(1 - a))
)

geo_orders[
    ["order_id", "seller_customer_distance_km"]
].head()
```

```python
geo_features = (
    geo_orders
    .groupby("order_id")
    .agg(
        avg_seller_customer_distance_km=(
            "seller_customer_distance_km",
            "mean"
        ),
        min_seller_customer_distance_km=(
            "seller_customer_distance_km",
            "min"
        ),
        max_seller_customer_distance_km=(
            "seller_customer_distance_km",
            "max"
        )
    )
    .reset_index()
)

geo_features.head()
```

```python
order_sellers = (
    order_items[
        ["order_id", "seller_id"]
    ]
    .drop_duplicates()
)

order_sellers = order_sellers.merge(
    order_base[
        [
            "order_id",
            "order_purchase_timestamp",
            "late_delivery"
        ]
    ],
    on="order_id",
    how="inner"
)

print("Order-seller pairs:", len(order_sellers))
order_sellers.head()
```

```python
order_sellers = order_sellers.sort_values(
    ["seller_id", "order_purchase_timestamp"]
)

order_sellers["seller_previous_orders"] = (
    order_sellers
    .groupby("seller_id")
    .cumcount()
)

order_sellers["seller_previous_lates"] = (
    order_sellers
    .groupby("seller_id")["late_delivery"]
    .cumsum()
    .shift(1)
    .fillna(0)
)
```

```python
order_sellers["seller_historical_late_rate"] = np.where(
    order_sellers["seller_previous_orders"] > 0,
    order_sellers["seller_previous_lates"]
    / order_sellers["seller_previous_orders"],
    np.nan
)
```

```python
seller_history_features = (
    order_sellers
    .groupby("order_id")
    .agg(
        avg_seller_historical_late_rate=(
            "seller_historical_late_rate",
            "mean"
        ),
        max_seller_historical_late_rate=(
            "seller_historical_late_rate",
            "max"
        ),
        avg_seller_previous_orders=(
            "seller_previous_orders",
            "mean"
        ),
        max_seller_previous_orders=(
            "seller_previous_orders",
            "max"
        )
    )
    .reset_index()
)
```

```python
seller_order_history = (
    order_items[
        ["order_id", "seller_id"]
    ]
    .drop_duplicates()
    .merge(
        delivered_orders[
            [
                "order_id",
                "order_purchase_timestamp",
                "order_delivered_customer_date",
                "late_delivery"
            ]
        ],
        on="order_id",
        how="inner"
    )
)

print("Seller-order pairs:", len(seller_order_history))
seller_order_history.head()
```

```python
seller_events = (
    seller_order_history[
        [
            "seller_id",
            "order_id",
            "order_delivered_customer_date",
            "late_delivery"
        ]
    ]
    .dropna(subset=["order_delivered_customer_date"])
    .sort_values(
        ["order_delivered_customer_date", "seller_id"]
    )
    .reset_index(drop=True)
)

seller_events["completed_orders_so_far"] = (
    seller_events
    .groupby("seller_id")
    .cumcount() + 1
)

seller_events["completed_lates_so_far"] = (
    seller_events
    .groupby("seller_id")["late_delivery"]
    .cumsum()
)
```

```python
current_seller_orders = (
    seller_order_history[
        [
            "order_id",
            "seller_id",
            "order_purchase_timestamp"
        ]
    ]
    .sort_values(
        ["order_purchase_timestamp", "seller_id"]
    )
    .copy()
)
```

```python
current_seller_orders = (
    seller_order_history[
        [
            "order_id",
            "seller_id",
            "order_purchase_timestamp"
        ]
    ]
    .sort_values(
        ["order_purchase_timestamp", "seller_id"]
    )
    .reset_index(drop=True)
)

current_seller_orders.head()
```

```python
print(
    "Missing delivery dates:",
    seller_events["order_delivered_customer_date"].isna().sum()
)

print(
    "Seller events:",
    len(seller_events)
)
```

```python
seller_history = pd.merge_asof(
    current_seller_orders,
    seller_events[
        [
            "seller_id",
            "order_delivered_customer_date",
            "completed_orders_so_far",
            "completed_lates_so_far"
        ]
    ],
    left_on="order_purchase_timestamp",
    right_on="order_delivered_customer_date",
    by="seller_id",
    direction="backward",
    allow_exact_matches=False
)

seller_history.head(20)
```
